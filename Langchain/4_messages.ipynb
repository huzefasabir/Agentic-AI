Messages

messages are the fundamental unit of context for models in Langchain. They represent the input and output of the models, carrying both the content and metadata needed to represent the state of a conversation whene interacting wiht an LLM. Messages are objects that contain 
  1) Roles: identifoes the message type (eg. system, user 
  2) Content : Represents the actual content if the messages(like text , images,audio, documents,etc)
  3) Metadata: optional fields such as response information, messages Ids, and token usage

  LAngchain provides a standard message typethat works across alll models provides, ensuring  consistent behavior regardless of the model being called.
